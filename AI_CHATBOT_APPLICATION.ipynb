```python
# =========================AI CHATBOT APPLICATION===================================
#
# Google Colab - Single Cell
# ============================================================

!pip -q install transformers torch gradio

import gradio as gr
from transformers import pipeline

# ------------------------------------------------------------
# Load Chatbot Model
# ------------------------------------------------------------

print("Loading chatbot model...")

chatbot = pipeline(
    "text-generation",
    model="microsoft/DialoGPT-small"
)

print("Chatbot model loaded successfully!")


# ------------------------------------------------------------
# Chat Function
# ------------------------------------------------------------

def chat(message, history):

    if not message.strip():
        return "Please enter a message."

    try:
        # Create conversation text
        conversation = ""

        for user_msg, bot_msg in history:
            conversation += f"User: {user_msg}\nBot: {bot_msg}\n"

        conversation += f"User: {message}\nBot:"

        result = chatbot(
            conversation,
            max_new_tokens=80,
            do_sample=True,
            temperature=0.7,
            pad_token_id=chatbot.tokenizer.eos_token_id
        )

        response = result[0]["generated_text"]

        # Extract chatbot response
        response = response.split("Bot:")[-1].strip()

        if "User:" in response:
            response = response.split("User:")[0].strip()

        return response

    except Exception as e:
        return f"Error: {str(e)}"


# ------------------------------------------------------------
# Gradio Chatbot Interface
# ------------------------------------------------------------

with gr.Blocks(title="AI Chatbot") as app:

    gr.Markdown(
        """
        # 🤖 AI Chatbot

        ### Simple NLP Conversational Chatbot

        Ask questions or have a conversation with the AI chatbot.
        """
    )

    chatbot_ui = gr.ChatInterface(
        fn=chat,
        title="💬 Chat with AI",
        textbox=gr.Textbox(
            placeholder="Type your message here...",
            container=True
        )
    )

    gr.Markdown(
        """
        **Technologies:** Python | NLP | Transformers | DialoGPT | Gradio | Google Colab
        """
    )


# ------------------------------------------------------------
# Launch Application
# ------------------------------------------------------------

app.launch(share=True)
```
